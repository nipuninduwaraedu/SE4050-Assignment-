```python
# ==========================================
# CELL 3: 1D CNN Training & Evaluation
# ==========================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Dropout, Dense, Flatten, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix

np.random.seed(42)
tf.random.set_seed(42)

# Data Loading (අවශ්‍ය නම් path එක වෙනස් කරන්න e.g., '/content/prepared_har_data.npz')
data_path = '../prepared_har_data.npz' if os.path.exists('../prepared_har_data.npz') else 'prepared_har_data.npz'
if not os.path.exists(data_path):
    data_path = '/content/prepared_har_data.npz'

data = np.load(data_path)

X_train, y_train = data['X_train'], data['y_train']
X_val, y_val     = data['X_val'], data['y_val']
X_test, y_test   = data['X_test'], data['y_test']

activity_names = data['activity_names'] if 'activity_names' in data else [
    'WALKING', 'WALKING_UPSTAIRS', 'WALKING_DOWNSTAIRS', 'SITTING', 'STANDING', 'LAYING'
]

# Model Architecture
input_shape = (X_train.shape[1], X_train.shape[2])
num_classes = len(np.unique(y_train))

model = Sequential([
    Conv1D(filters=64, kernel_size=5, activation='relu', input_shape=input_shape, padding='same'),
    BatchNormalization(),
    MaxPooling1D(pool_size=2),
    Dropout(0.3),

    Conv1D(filters=128, kernel_size=3, activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling1D(pool_size=2),
    Dropout(0.3),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(num_classes, activation='softmax')
], name="1D_CNN_HAR")

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

checkpoint_path = 'saved_models/model2_1d_cnn.h5'
callbacks = [
    EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True, verbose=1),
    ModelCheckpoint(checkpoint_path, monitor='val_accuracy', save_best_only=True, verbose=1)
]

start_time = time.time()

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=64,
    callbacks=callbacks,
    verbose=1
)

training_time_seconds = time.time() - start_time

# Evaluation
best_model = tf.keras.models.load_model(checkpoint_path)
test_loss, test_accuracy = best_model.evaluate(X_test, y_test, verbose=0)
y_pred_probs = best_model.predict(X_test)
y_pred = np.argmax(y_pred_probs, axis=1)

precision, recall, f1, _ = precision_recall_fscore_support(y_test, y_pred, average='macro')

# Save Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=activity_names, yticklabels=activity_names)
plt.title('1D CNN - Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.tight_layout()
plt.savefig('results/model2/confusion_matrix.png', dpi=300)
plt.close()

# Save Learning Curves
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(history.history['accuracy'], label='Train Accuracy')
ax1.plot(history.history['val_accuracy'], label='Val Accuracy')
ax1.set_title('Model Accuracy')
ax1.legend()

ax2.plot(history.history['loss'], label='Train Loss')
ax2.plot(history.history['val_loss'], label='Val Loss')
ax2.set_title('Model Loss')
ax2.legend()

plt.tight_layout()
plt.savefig('results/model2/learning_curves.png', dpi=300)
plt.close()

# Save Metrics CSV
trainable_params = sum([tf.keras.backend.count_params(w) for w in model.trainable_weights])
metrics_df = pd.DataFrame([{
    'Model': '1D CNN',
    'Test Accuracy': test_accuracy,
    'Precision (Macro)': precision,
    'Recall (Macro)': recall,
    'F1-Score (Macro)': f1,
    'Training Time (s)': training_time_seconds,
    'Trainable Parameters': trainable_params
}])
metrics_df.to_csv('results/model2/metrics.csv', index=False)

print("\n--- Training and Evaluation Completed Successfully! ---")
```

```python
# ==========================================
# CELL 3: 1D CNN Training & Evaluation (Direct .npy Loader)
# ==========================================
import os
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Dropout, Dense, Flatten, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix

np.random.seed(42)
tf.random.set_seed(42)

# Direct .npy Files Loading
print("Loading individual .npy files...")
X_train = np.load('X_train.npy')
y_train = np.load('y_train.npy')

# Validation set එක X_val/X_validation ලෙස පැවතිය හැක
if os.path.exists('X_val.npy'):
    X_val = np.load('X_val.npy')
    y_val = np.load('y_val.npy')
else:
    X_val = np.load('X_validation.npy')
    y_val = np.load('y_validation.npy')

X_test = np.load('X_test.npy')
y_test = np.load('y_test.npy')

if os.path.exists('activity_names.npy'):
    activity_names = np.load('activity_names.npy', allow_pickle=True)
else:
    activity_names = ['WALKING', 'WALKING_UPSTAIRS', 'WALKING_DOWNSTAIRS', 'SITTING', 'STANDING', 'LAYING']

print(f"Data Loaded Successfully! X_train shape: {X_train.shape}")

# Model Architecture
input_shape = (X_train.shape[1], X_train.shape[2])
num_classes = len(np.unique(y_train))

model = Sequential([
    Conv1D(filters=64, kernel_size=5, activation='relu', input_shape=input_shape, padding='same'),
    BatchNormalization(),
    MaxPooling1D(pool_size=2),
    Dropout(0.3),

    Conv1D(filters=128, kernel_size=3, activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling1D(pool_size=2),
    Dropout(0.3),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(num_classes, activation='softmax')
], name="1D_CNN_HAR")

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

os.makedirs('saved_models', exist_ok=True)
os.makedirs('results/model2', exist_ok=True)

checkpoint_path = 'saved_models/model2_1d_cnn.h5'
callbacks = [
    EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True, verbose=1),
    ModelCheckpoint(checkpoint_path, monitor='val_accuracy', save_best_only=True, verbose=1)
]

start_time = time.time()

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=64,
    callbacks=callbacks,
    verbose=1
)

training_time_seconds = time.time() - start_time

# Evaluation
best_model = tf.keras.models.load_model(checkpoint_path)
test_loss, test_accuracy = best_model.evaluate(X_test, y_test, verbose=0)
y_pred_probs = best_model.predict(X_test)
y_pred = np.argmax(y_pred_probs, axis=1)

precision, recall, f1, _ = precision_recall_fscore_support(y_test, y_pred, average='macro')

# Save Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=activity_names, yticklabels=activity_names)
plt.title('1D CNN - Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.tight_layout()
plt.savefig('results/model2/confusion_matrix.png', dpi=300)
plt.close()

# Save Learning Curves
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(history.history['accuracy'], label='Train Accuracy')
ax1.plot(history.history['val_accuracy'], label='Val Accuracy')
ax1.set_title('Model Accuracy')
ax1.legend()

ax2.plot(history.history['loss'], label='Train Loss')
ax2.plot(history.history['val_loss'], label='Val Loss')
ax2.set_title('Model Loss')
ax2.legend()

plt.tight_layout()
plt.savefig('results/model2/learning_curves.png', dpi=300)
plt.close()

# Save Metrics CSV
trainable_params = sum([tf.keras.backend.count_params(w) for w in model.trainable_weights])
metrics_df = pd.DataFrame([{
    'Model': '1D CNN',
    'Test Accuracy': test_accuracy,
    'Precision (Macro)': precision,
    'Recall (Macro)': recall,
    'F1-Score (Macro)': f1,
    'Training Time (s)': training_time_seconds,
    'Trainable Parameters': trainable_params
}])
metrics_df.to_csv('results/model2/metrics.csv', index=False)

print("\n--- Training and Evaluation Completed Successfully! ---")
```
